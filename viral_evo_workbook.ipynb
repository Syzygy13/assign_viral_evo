````python
need_for_JC_correction = """
The raw distance does not account for the actual number of substitutions as bases can mutate more than once. These bases look like only one
change occurred, or none in the case the base mutated back to its original form. As time passes there are more of these hidden mutations.
Two sequences would match roughly 25'%' of the time by pure chance as there are only four bases. Therefore their Hamming distance will cap
at apporximately 75'%'. This is fixed through JC correction as it estimates the true number of substitutions per base using the formula 
d = (3/4) * ln(1 - (4/3)p). This makes distance proportional to the amount of time passed which is assumed by a linear regression for the
mutation. SARS-CoV-2 sequences are similar so it is not as noticeable, however more diverged viruses would show more different results.
"""
````

````python
# Essentially the same as problem 1
def estimate_mutation_rates(fasta_path):
    genomes = []
    for record in SeqIO.parse(fasta_path, "fasta"):
        accession, country, collection_date = record.description.split("|")
        genomes.append({
            "accession": accession,
            "country": country,
            "collection_date": pd.to_datetime(collection_date, format="mixed"),
            "sequence": str(record.seq),
        })
    genomes = pd.DataFrame(genomes)

    # Get the reference genome again
    ref_seq = genomes["sequence"].iloc[0]

    # Make 2 new columns for the date since the reference and the distance using JC-correction
    earliest = genomes["collection_date"].iloc[0]
    genomes["day_count"] = (genomes["collection_date"] - earliest).dt.days
    genomes["distance"] = [jukes_cantor(ref_seq, seq) for seq in genomes["sequence"]]

    # Make the line based on every genome except the reference
    other_genomes = genomes.iloc[1:]
    slope_genome, intercept = np.polyfit(other_genomes["day_count"], other_genomes["distance"], deg=1)

    # Determine the slope for mutations per nucleotide
    no_gap_genome = len(ref_seq.replace("-", ""))
    slope_nt = slope_genome / no_gap_genome

    return slope_genome, slope_nt, genomes, slope_genome, intercept

# Plot everything, close to before again
def plot_distance_over_time(genomes, slope, intercept, virus_name):
    # Determine the use of a scatterplot
    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.scatter(genomes["day_count"], genomes["distance"], color="blue", label=f"{virus_name} genomes")
    
    # Mark the reference
    ax.scatter([0], [0], color="purple", label="Reference")

    # Fill the rest of the plot with the genomes
    fit_days = np.array([0, genomes["day_count"].max()])
    ax.plot(fit_days, slope * fit_days + intercept, color="red",
            label=f"fit: {slope:.4f} mutations/genome/day")

    # Basic plot settings
    ax.set_xlabel("Days after reference sample")
    ax.set_ylabel("JC-corrected distance (mutations per genome)")
    ax.set_title(f"{virus_name}: genetic distance from reference over time")
    ax.legend()
    fig.tight_layout()

    return fig

# Get data for plots and final answers
ebola_per_genome_per_day, ebola_per_nt_per_day, ebola_genomes, ebola_slope, ebola_intercept = estimate_mutation_rates("data/p2-ebola-viruses.fasta")
zika_per_genome_per_day, zika_per_nt_per_day, zika_genomes, zika_slope, zika_intercept = estimate_mutation_rates("data/p2-zika-viruses.fasta")

plot_distance_over_time(ebola_genomes, ebola_slope, ebola_intercept, "Ebola").savefig("problem2_ebola.svg")
plot_distance_over_time(zika_genomes, zika_slope, zika_intercept, "Zika").savefig("problem2_zika.svg")
plt.show()
````

## Problem 3: Reinterpreting mutation rates

Majority of RNA viral genomes are single-stranded RNA, where only one strand of the viruses encode for the genes. We divide these viruses into positive-sense and negative-sense viruses, based on where they code for genes.
Positive-sense RNA directly codes for genes and acts like mRNA, meaning it can be directly used by the host's ribosomes to make proteins.
Negative-sense RNA is a reverse complement of the positive-sense RNA and does not code for genes directly. It must first be converted into positive-sense RNA by the virus's own enzymes before it can be used to make proteins.
Positive-sense RNA viruses mutate faster because their replication process is simpler and mostly lacks error correction. Negative-sense RNA viruses, on the other hand, mutate more slowly since they have an extra step in replication and require more functional genes to support replication.

## Problem 4: Variant-specific mutations

As the virus mutates, it inevitably evolves and proliferates around the world. Every so often, some mutations may prove especially beneficial to the spread of the virus, and this version of the virus spreads faster than other versions. When a version of a virus becomes especially prevalent inside a population, we call this a virus *variant*. Variants are nothing more than a naming scheme for viruses that have specific mutations. For instance, in Slovenia during 2022, we were dealing with the Omicron variant. Think of this as observing natural selection in real time. Some viruses have mutations that enable them to spread more easily throughout our population, which inevitably leads to the demise of other virus variants, which are not as good at proliferation. The result is survival of the fittest at the viral level, where, unfortunately, the fittest viruses seem to cause the most damage to us humans.

How do we identify variants? A variant is determined by several so-called *defining mutations*. Mutations can either be synonymous or nonsynonymous. Synonymous mutations are changes in nucleotide bases that result in the same encoded amino acid and are thus less important. Nonsynonymous mutations are nucleotide mutations that alter the amino acid sequence of a protein.

````python
# Function to get the frequency of mutations
def mutations(accessions, genomes, reference):
    # Get all the genomes from list and add to a table
    genome_rows = []
    for accession in accessions:
        genome_rows.append(genomes[accession])
    all_genomes = np.array(genome_rows)

    # Determine the number of mutations and not ambiguous bases
    is_mutated = all_genomes != reference
    not_ambiguous = all_genomes != "N"
    mutated_num = (is_mutated & not_ambiguous).sum(axis=0)
    total_num = np.maximum(not_ambiguous.sum(axis=0), 1)

    # Return the frequency of mutations
    return mutated_num / total_num

# Function to find positions that are mutated a lot
def vital_sites(frequency, gene):
    # Find positions that are mutated in more than half the genomes
    gene_start, gene_end = gene_locations[gene]
    vital_sites = []
    for position, freq in zip(positions, frequency):
        inside_gene = gene_start <= position <= gene_end
        if inside_gene and freq > 0.5:
            vital_sites.append(int(position))
    return vital_sites

# Function to find the most common nucleotide for a position in the set of genomes
def most_common_base(accessions, genomes, site):
    # Store the amount of each base at each position in all genomes in a dictionary
    base_counts = {}
    for accession in accessions:
        # The base at the required location
        base = genomes[accession][site - 1]
        if base in base_counts:
            base_counts[base] += 1
        else:
            base_counts[base] = 1

    # Determine the most common base
    best_base = None
    best_count = 0
    for base, count in base_counts.items():
        if count > best_count:
            best_base = base
            best_count = count
    return best_base

# Gather all sequences and store in an array
genomes = {}
for record in SeqIO.parse("data/p1-sars-cov-2-variants.fasta", "fasta"):
    accession = record.id.split("|")[0]
    letters = list(str(record.seq).upper())
    genomes[accession] = np.array(letters)

# Store the reference and positions of its nucleotides
reference = genomes["NC_045512.2"]
positions = np.arange(1, len(reference) + 1)

# Find the frequency of mutations in the alpha and delta variants
freq_alpha = mutations(alpha_variants, genomes, reference)
freq_delta = mutations(delta_variants, genomes, reference)

# --- Plot: Alpha on top, Delta below, only the genome past position 20,000 ---
colors = {
    "S": "red",
    "E": "blue",
    "M": "green",
    "N": "purple"
}

# Plot the 2 results on the same image. Twice for alpha and delta
fig, axes = plt.subplots(2, 1, figsize=(11, 6), sharex=True, sharey=True)
for ax, freq, name in [(axes[0], freq_alpha, "Alpha (20I)"), (axes[1], freq_delta, "Delta (21A)")]:
    ax.plot(positions[19999:], freq[19999:])
    # For each segment of the genes
    for gene, (start, end) in gene_locations.items():
        ax.axvspan(start, end, color=colors[gene], alpha=0.2)
        ax.text((start + end) / 2, 1.1, gene, color=colors[gene])
    ax.set_ylabel("Mutation frequency")
    ax.set_title(name)
    ax.set_ylim(0, 1.2)

# Basic plot settings
axes[1].set_xlabel("Position in reference genome (nt)")
fig.suptitle("Mutation frequency along the SARS-CoV-2 genome (positions > 20,000)")
fig.savefig("problem3.svg")
plt.show()

# Get the areas in the spike region with > 50% mutation rate
alpha_vital = vital_sites(freq_alpha, "S")
delta_vital = vital_sites(freq_delta, "S")

# Combine the mutations in both into one list
mutations_combined = []
for pos in (alpha_vital + delta_vital):
    mutations_combined.append(int(pos))
mutations_combined.sort()

# Find mutations that occur in both alpha and delta variants
mutations_in_both_variants = []
for index in range(1, len(mutations_combined)):
    if mutations_combined[index] == mutations_combined[index - 1]:
        mutations_in_both_variants.append(mutations_combined[index])

# Find the vital site where both variants changed to the same nucleotide
for site in mutations_combined:
    alpha_base = most_common_base(alpha_variants, genomes, site)
    delta_base = most_common_base(delta_variants, genomes, site)
    
    # Check if they are the same and set the answer as that
    if alpha_base == delta_base:
        same_mutation = f"{reference[site - 1]}{site}{alpha_base}"
        print(same_mutation, site)
````

````python
from Bio.Seq import Seq

# Function to find if a sequence matches the reference exactly
def contains_mutation(spike_sequence, ref_spike):
    for base, expected in zip(spike_sequence, ref_spike):
        # If an X is not found and it doesn't match it returns True
        if base != "X" and base != expected:
            return True
    return False

# Function to determine percentage of mutations in sequence
def match_percentage(spike_sequence, spike_mutations):
    matches = 0
    for mutation in spike_mutations:

        # Find the correctly aligned base
        base = spike_sequence[mutation["pos"] - 1]
        if base == mutation["right"]:
            matches += 1

    # Return as percentage
    return (matches / len(spike_mutations)) * 100

# Function to classify the sequence
def which_type(spike_sequence, ref_spike, variant_spikes):
    # If there are no mutations mark it as such
    if not contains_mutation(spike_sequence, ref_spike):
        return "NO_MUTATIONS"

    # Determine the best match based on the 51% required
    best = "UNKNOWN"
    best_percent = 51

    # Compare percentages several times to get the best variant
    for name, spike_mutations in variant_spikes.items():
        percentage = match_percentage(spike_sequence, spike_mutations)
        if percentage > best_percent:
            best = name
            best_percent = percentage
    return best

# Find the reference genome once more
ref = ""
for record in SeqIO.parse("data/p1-sars-cov-2-variants.fasta", "fasta"):
    if record.id.split("|")[0] == "NC_045512.2":
        ref = str(record.seq).replace("-", "")

# Find the spike protein based on the bases and its length
spike_start = ref.find("ATGTTTGTTTTTCTTGTTTTATTGCC")
ref_spike = str(Seq(ref[spike_start:spike_start + 1273 * 3]).translate())

# Open the variants.json file for use
variants = pd.read_json("data/variants.json", typ="series")

variant_spikes = {}
for name, data in variants.items():
    spike_mutations = []

    # Find spike proteins that are nonsynonymous only
    for mutation in data["nonsynonymous"]:
        if mutation["gene"] == "S":
            spike_mutations.append(mutation)

    # Store all the spikes for each variant
    variant_spikes[name] = spike_mutations
    
# Use the problem 5 file to parse through variants
rows = []
for record in SeqIO.parse("data/p5-aus-spike-proteins.fasta", "fasta"):
    name, date, accession = record.description.split("|")
    rows.append({
        "accession": accession,
        "date": pd.to_datetime(date),
        "variant": which_type(str(record.seq).upper(), ref_spike, variant_spikes),
    })
typed = pd.DataFrame(rows)

# Add them all to the csv file
typed[["accession", "variant"]].to_csv("problem5-classification.csv", sep=",", header=False, index=False)

# Group the variants by the month they occurred and the variants themselves
num_per_month = typed.groupby([pd.Grouper(key="date", freq="MS"), "variant"]).size()
num_per_month = num_per_month.unstack(fill_value=0)

# Determine the percentage per month using the table made
percent_per_month = num_per_month.div(num_per_month.sum(axis=1), axis=0) * 100

# Format the column orders so NO_MUTATIONS and UNKNOWN come after the other variants
column_order = []
for name in variant_spikes:
    if name in percent_per_month.columns:
        column_order.append(name)
for mutation in ["NO_MUTATIONS", "UNKNOWN"]:
    if mutation in percent_per_month.columns:
        column_order.append(mutation)
percent_per_month = percent_per_month[column_order]

# Give a colour to each variant using a colour map
colour_palette = plt.get_cmap("tab20").colors
colours = []
for i, class_name in enumerate(column_order):
    # Specify the colours for NO_MUTATIONS and UNKNOWN
    if class_name == "NO_MUTATIONS":
        colours.append("white")
    elif class_name == "UNKNOWN":
        colours.append("lightgrey")
    else:
        colours.append(colour_palette[i % len(colour_palette)])

# Plot the entire thing using basic settings as before
fig, ax = plt.subplots(figsize=(12, 5.5))
ax.stackplot(percent_per_month.index, percent_per_month.T.values, labels=column_order, colors=colours)
ax.set_ylim(0, 100)
ax.set_xlim(percent_per_month.index.min(), percent_per_month.index.max())
ax.set_xlabel("Collection month")
ax.set_ylabel("Percentage of sequences (%)")
ax.set_title("SARS-CoV-2 variants in Australia over time classified by spikes")
ax.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), title="Variants")
fig.savefig("problem5.svg", bbox_inches="tight")
plt.show()
````
